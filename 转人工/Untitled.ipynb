```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

# 设置中文字体
plt.rcParams["font.family"] = ["SimHei", "WenQuanYi Micro Hei", "Heiti TC"]
plt.rcParams['axes.unicode_minus'] = False

# 原始数据
data = {
    '人群标签': ['L1', 'L2', 'L3', 'L4', 'L5', '未知'],
    'R1': [989, 1081, 886, 362, 115, 0],
    'R2': [1126, 2473, 3206, 2667, 1051, 0],
    'R3': [347, 1305, 2006, 2396, 1262, 0],
    'R4': [13, 180, 616, 1901, 3100, 0],
    '未知': [2021, 4422, 4579, 4629, 3384, 287]
}

df = pd.DataFrame(data)

# 计算R到L的转移概率矩阵
def calculate_transition_matrix(df):
    r_levels = ['R1', 'R2', 'R3', 'R4']
    l_levels = ['L1', 'L2', 'L3', 'L4', 'L5', '未知']
    T = np.zeros((len(r_levels), len(l_levels)))
    for i, r in enumerate(r_levels):
        total_r = df[r].sum()
        for j, l in enumerate(l_levels):
            # df 中行是 L 水平，列是 R 水平
            T[i, j] = df.loc[df['人群标签'] == l, r].values[0] / total_r if total_r > 0 else 0
    return T, r_levels, l_levels

# 计算当前分布（用于对比，但在主流程中不参与链仿真）
def calculate_current_distribution(df):
    l_levels = ['L1', 'L2', 'L3', 'L4', 'L5', '未知']
    dist = {}
    for l in l_levels:
        # 汇总该 L 级别下所有 R1–R4 的人数
        dist[l] = df.loc[df['人群标签'] == l, ['R1','R2','R3','R4']].sum().sum()
    return dist

# 修改转移矩阵以实现不同策略
def modify_transition_matrix(T, strategy, r_levels, l_levels):
    M = T.copy()
    r4_idx = r_levels.index('R4')
    l4_idx = l_levels.index('L4')
    l5_idx = l_levels.index('L5')
    if strategy == 1:  # R4→L4 和 L5，按原有比例分配
        total = T[r4_idx, l4_idx] + T[r4_idx, l5_idx]
        M[r4_idx, :] = 0
        if total > 0:
            M[r4_idx, l4_idx] = T[r4_idx, l4_idx] / total
            M[r4_idx, l5_idx] = T[r4_idx, l5_idx] / total
    elif strategy == 2:  # R4→纯 L5
        M[r4_idx, :] = 0
        M[r4_idx, l5_idx] = 1.0
    return M

# 模拟马尔可夫链
def simulate_markov_chain(T, initial_state, steps=10):
    """
    T: 转移矩阵，shape = (4,6)，行是当前 R1–R4，列是下步 L1–L5、未知
    initial_state: 长度 4 的数组，表示 R1–R4 的初始比例
    返回：shape = (steps+1, 6) 的数组，表示每一步 L1–L5、未知 分布
    """
    state = initial_state.copy()
    history = [state.dot(T)]  # 第一步从 R→L
    for _ in range(1, steps):
        # 后续如果想基于同一维度更新，可以将 L 级别重新映射到 R，再迭代；这里仅演示一次映射
        history.append(history[-1])  # 如果要多步迭代，需要定义 L→R 的映射矩阵
    return np.array(history)

# 计算KL散度
def kl_divergence(p, q):
    p = np.array(p) + 1e-10
    q = np.array(q) + 1e-10
    p /= p.sum()
    q /= q.sum()
    return np.sum(p * np.log(p / q))

def main():
    # 1. 计算原始转移矩阵
    T, r_levels, l_levels = calculate_transition_matrix(df)
    print("原始转移矩阵：")
    print(pd.DataFrame(T, index=r_levels, columns=l_levels))

    # 2. 计算并展示策略矩阵
    M1 = modify_transition_matrix(T, 1, r_levels, l_levels)
    M2 = modify_transition_matrix(T, 2, r_levels, l_levels)
    print("\n策略1（R4→L4+L5）矩阵：")
    print(pd.DataFrame(M1, index=r_levels, columns=l_levels))
    print("\n策略2（R4→L5）矩阵：")
    print(pd.DataFrame(M2, index=r_levels, columns=l_levels))

    # 3. 构造 R 层级的初始分布
    r_counts = np.array([df[r].sum() for r in r_levels])
    initial_state = r_counts / r_counts.sum()

    # 4. 模拟一次 R→L，比较三种矩阵下的 L 分布
    steps = 1  # 只做一次映射
    hist_orig = simulate_markov_chain(T, initial_state, steps)
    hist_1 = simulate_markov_chain(M1, initial_state, steps)
    hist_2 = simulate_markov_chain(M2, initial_state, steps)

    final_orig = hist_orig[-1]
    final_1 = hist_1[-1]
    final_2 = hist_2[-1]

    print("\n原始策略下 L 分布：", dict(zip(l_levels, final_orig)))
    print("策略1 下 L 分布：", dict(zip(l_levels, final_1)))
    print("策略2 下 L 分布：", dict(zip(l_levels, final_2)))

    # 5. KL 散度比较
    kl1 = kl_divergence(final_orig, final_1)
    kl2 = kl_divergence(final_orig, final_2)
    print(f"\n策略1 与原始的 KL 散度：{kl1:.6f}")
    print(f"策略2 与原始的 KL 散度：{kl2:.6f}")

    # 6. 可视化（如需多步迭代，可自行扩展 simulate_markov_chain）
    plt.figure(figsize=(8,6))
    x = np.arange(len(l_levels))
    width = 0.25
    plt.bar(x-width, final_orig, width, label='原始')
    plt.bar(x,      final_1,  width, label='策略1')
    plt.bar(x+width,final_2,  width, label='策略2')
    plt.xticks(x, l_levels)
    plt.xlabel('L 级别')
    plt.ylabel('比例')
    plt.title('一次映射后的 L 分布对比')
    plt.legend()
    plt.tight_layout()
    plt.show()

if __name__ == "__main__":
    main()
```

```python
import numpy as np
from scipy.stats import entropy
import pandas as pd

# Create the data matrix
data = np.array([
    [989, 1126, 347, 13, 2021],
    [1081, 2473, 1305, 180, 4422], 
    [886, 3206, 2006, 616, 4579],
    [362, 2667, 2396, 1901, 4629],
    [115, 1051, 1262, 3100, 3384]
])

# Original distribution
orig_dist = data / data.sum()

# Strategy 1: R4 to L4 and L5
strat1_data = data.copy()
r4_total = data[:, 3].sum()
strat1_data[3, 3] += r4_total * 0.4  # 40% to L4
strat1_data[4, 3] += r4_total * 0.6  # 60% to L5
strat1_dist = strat1_data / strat1_data.sum()

# Strategy 2: R4 directly to L5
strat2_data = data.copy() 
strat2_data[4, 3] += r4_total  # 100% to L5
strat2_dist = strat2_data / strat2_data.sum()

# Calculate KL divergence
kl_strat1 = entropy(orig_dist.flatten(), strat1_dist.flatten())
kl_strat2 = entropy(orig_dist.flatten(), strat2_dist.flatten())

print("KL Divergence Results:")
print(f"Strategy 1 (R4 to L4+L5): {kl_strat1:.6f}")
print(f"Strategy 2 (R4 to L5): {kl_strat2:.6f}")

# Monte Carlo simulation
n_simulations = 10000
kl_sim1 = []
kl_sim2 = []

for _ in range(n_simulations):
    # Add random noise
    noise = np.random.normal(0, 0.01, data.shape)
    noisy_data = data + noise * data
    noisy_data[noisy_data < 0] = 0
    
    noisy_dist = noisy_data / noisy_data.sum()
    
    # Apply strategies with noise
    sim1_data = noisy_data.copy()
    sim1_data[3, 3] += r4_total * 0.4
    sim1_data[4, 3] += r4_total * 0.6
    sim1_dist = sim1_data / sim1_data.sum()
    
    sim2_data = noisy_data.copy()
    sim2_data[4, 3] += r4_total
    sim2_dist = sim2_data / sim2_data.sum()
    
    kl_sim1.append(entropy(noisy_dist.flatten(), sim1_dist.flatten()))
    kl_sim2.append(entropy(noisy_dist.flatten(), sim2_dist.flatten()))

print("\nMonte Carlo Simulation Results:")
print(f"Strategy 1 mean KL: {np.mean(kl_sim1):.6f} ± {np.std(kl_sim1):.6f}")
print(f"Strategy 2 mean KL: {np.mean(kl_sim2):.6f} ± {np.std(kl_sim2):.6f}")
```
